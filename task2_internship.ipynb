```python
# Step 1: Install libraries, since i am using colab
!pip install yfinance scikit-learn

# Step 2: Import libraries
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Step 3: Load stock data (example: Apple)
stock = yf.download('AAPL', start='2022-01-01', end='2024-01-01')

# Step 4: Prepare dataset
# Use Open, High, Low, Volume as input features
X = stock[['Open', 'High', 'Low', 'Volume']]

# Target = Next day's Close price
y = stock['Close'].shift(-1)

# Remove last row (NaN due to shift)
X = X[:-1]
y = y[:-1]

# Step 5: Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# Step 6: Train Model (Linear Regression)
model = LinearRegression()
model.fit(X_train, y_train)

# Step 7: Predict
y_pred = model.predict(X_test)

# Step 8: Plot Actual vs Predicted
plt.figure(figsize=(10,5))
plt.plot(y_test.values, label='Actual Price')
plt.plot(y_pred, label='Predicted Price')
plt.title('Actual vs Predicted Closing Price')
plt.xlabel('Days')
plt.ylabel('Price')
plt.legend()
plt.show()
```
